```python
import joblib
import numpy as np
import tensorflow as tf

# Load trained model
model = tf.keras.models.load_model("models/priority_model.keras")

# Load scaler
scaler = joblib.load("models/scaler.pkl")


def predict_priority(complaint, safety_risk):

    features = np.array([[
        complaint["severity"],
        complaint["traffic_impact"],
        complaint["public_impact"],
        complaint["weather_risk"],
        complaint["days_pending"],
        complaint["complaint_frequency"],
        safety_risk
    ]])

    # Scale using the same scaler used during training
    features = scaler.transform(features)

    # Predict priority
    priority_score = model.predict(features, verbose=0)[0][0]

    return float(np.clip(priority_score, 0, 1))


## Temporarily checking // remove this 
if __name__ == "__main__":

    complaint = {
        "severity": 8,
        "traffic_impact": 7,
        "public_impact": 6,
        "weather_risk": 5,
        "days_pending": 10,
        "complaint_frequency": 4
    }

    safety_risk = 0.6

    priority = predict_priority(complaint, safety_risk)

    print("Priority Score:", priority)
```
